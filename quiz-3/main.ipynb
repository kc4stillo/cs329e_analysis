```python
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
import pandas as pd

train_df = pd.read_csv("q3-train.csv")
test_df = pd.read_csv("q3-test.csv")

X_train = pd.DataFrame({ 'Feature1': train_df['Feature1'], 'Feature2': train_df['Feature2']})
y_train = pd.DataFrame({ 'Label': train_df['Label']})

X_test = pd.DataFrame({ 'Feature1': test_df['Feature1'], 'Feature2': test_df['Feature2']})
y_test = pd.DataFrame({ 'Label': test_df['Label']})

model = LogisticRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

score = accuracy_score(y_test, y_pred)

print("Accuracy Score: " + str(score))

```

```python
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC

train_df = pd.read_csv("q3-train.csv")
test_df = pd.read_csv("q3-test.csv")

X_train = pd.DataFrame({ 'Feature1': train_df['Feature1'], 'Feature2': train_df['Feature2']})
y_train = pd.DataFrame({ 'Label': train_df['Label']})

X_test = pd.DataFrame({ 'Feature1': test_df['Feature1'], 'Feature2': test_df['Feature2']})
y_test = pd.DataFrame({ 'Label': test_df['Label']})

model = SVC()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

score = accuracy_score(y_test, y_pred)

print("Accuracy Score: " + str(score))
```

```python
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier

train_df = pd.read_csv("q3-train.csv")
test_df = pd.read_csv("q3-test.csv")

X_train = pd.DataFrame({ 'Feature1': train_df['Feature1'], 'Feature2': train_df['Feature2']})
y_train = pd.DataFrame({ 'Label': train_df['Label']})

X_test = pd.DataFrame({ 'Feature1': test_df['Feature1'], 'Feature2': test_df['Feature2']})
y_test = pd.DataFrame({ 'Label': test_df['Label']})

model = KNeighborsClassifier(n_neighbors=100)
# KNeighborsClassifier?
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

score = accuracy_score(y_test, y_pred)

print("Accuracy Score: " + str(score))
```

```python
from sklearn.metrics import accuracy_score
# from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

train_df = pd.read_csv("q3-train.csv")
test_df = pd.read_csv("q3-test.csv")

X_train = pd.DataFrame({ 'Feature1': train_df['Feature1'], 'Feature2': train_df['Feature2']})
y_train = pd.DataFrame({ 'Label': train_df['Label']})

X_test = pd.DataFrame({ 'Feature1': test_df['Feature1'], 'Feature2': test_df['Feature2']})
y_test = pd.DataFrame({ 'Label': test_df['Label']})

model = DecisionTreeClassifier()
# KNeighborsClassifier?
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

score = accuracy_score(y_test, y_pred)

print("Accuracy Score: " + str(score))
```

```python
from sklearn.metrics import accuracy_score
from sklearn.ensemble import AdaBoostClassifier

train_df = pd.read_csv("q3-train.csv")
test_df = pd.read_csv("q3-test.csv")

X_train = pd.DataFrame({ 'Feature1': train_df['Feature1'], 'Feature2': train_df['Feature2']})
y_train = pd.DataFrame({ 'Label': train_df['Label']})

X_test = pd.DataFrame({ 'Feature1': test_df['Feature1'], 'Feature2': test_df['Feature2']})
y_test = pd.DataFrame({ 'Label': test_df['Label']})

model = AdaBoostClassifier()
# KNeighborsClassifier?
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

score = accuracy_score(y_test, y_pred)

print("Accuracy Score: " + str(score))
```
