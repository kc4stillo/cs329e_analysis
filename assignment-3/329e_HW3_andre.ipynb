# Question 4 
Formally test whether or not participants consumed 
more calories than non-participants at the $\alpha = 0.05$ level using the 5 steps procedure for hypothesis tests. **(6 points )**

Our hypothesis:
H_0: The participant mean calories is larger than the non partcipant mean calories
H_\alpha: The particpant mean calories is the same or smaller than the non participant mean calories

Our test:
We need to use a t-test because our sample size is smaller than 30 and our population standard deviation is unknown.

```python
from scipy import stats 

t_stat, p_val = stats.ttest_ind(
    a = pd.read_csv('participants.csv'),
    b = pd.read_csv('nonparticipants.csv'),
    alternative='less',
    equal_var=False
)

alpha = 0.05

print("p-val: " + str(p_val))

if p_val < alpha: 
    print("The null hypothesis is rejected because the p-val from the two sample t-test is smaller than tha alpha.")

elif p_val >= alpha:
    print("The null hypothesis was not rejected because the p-val from the two sample t-test was the same or larger than the two sample t-test alpha value.")
```
